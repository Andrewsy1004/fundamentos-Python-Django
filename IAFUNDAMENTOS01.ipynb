```python
# Importando Librerias
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
edades_colombia = np.array([10,20,30,50,70,100,56])

print(f"Edades en colombia:\n{edades_colombia}\n")

print(f"La edad máxima es: {edades_colombia.max()}")
print(f"La edad minima es: {edades_colombia.min()}")
print(f"La edad promedio en colombia es: {edades_colombia.mean()}")
print(f"La desviacion estandar: {edades_colombia.std():1f}")
```

```python
ventas_semana = np.array([
    [10,10,10,50,90,200,0],
    [10,10,10,50,90,300,0],
    [10,10,10,50,90,400,0],
    [10,10,10,50,90,500,0]
])

print(f"Ventas semana:\n{ventas_semana}\n")

print(f"La venta maxima es: {ventas_semana.max()}")

# toal de ventas por semanas
print(f"Total de ventas por semana: {ventas_semana.sum(axis=1)}")

# Maximo de ventas por semana
print(f"Maximo de ventas por semana: {ventas_semana.max(axis=1)}")

# Minimo de ventas por semana
print(f"Minimo de ventas por semana: {ventas_semana.min(axis=1)}")
```

```python
# DataFrames
datos_empresa = {
    "Id" : np.random.randint(1,100,10),
    "Nombre": np.random.choice(["Juan","Pedro","Maria","Andres","Erika"],10),
    "Ingresos": np.random.randint(1000,10000,10),
    "Gasto": np.random.randint(500,5000,10),
    "Genero": np.random.choice(["M","F"],10)
}

df = pd.DataFrame(datos_empresa)
print(df)
```

```python
# EDA ( Exploratory Data Analysis )

# si el nombre es nulo que lo ponga vacio ""
df["Nombre"] = df["Nombre"].fillna("")

# si los gastos e ingresos son menores a 0 colocar 0 por default
df["Ingresos"] = df["Ingresos"].apply(lambda x: 0 if x < 0 else x)
df["Gasto"] = df["Gasto"].apply(lambda x: 0 if x < 0 else x)

# Elimando a los hombres
# df = df[df["Genero"] != "M"]
```

```python
# Resumen Estadistico
df.describe()
```

```python
# Diagrama de barras que muestres, cuantas personas hay de sexo m y f

genero_pd = df["Genero"].value_counts()
genero_pd_columns = ["Genero","Cantidad"]

plt.figure(figsize=(6,4))

sns.barplot(x=genero_pd.index,y=genero_pd.values, palette=["blue","pink"])
plt.title("Cantidad de personas por genero")
plt.xlabel("Genero")
plt.ylabel("Cantidad")
plt.show()
```

```python
# Comparar cantidad de dinero que gana una mujer y un hombre

genero_pagos         = df.groupby("Genero")["Ingresos"].sum()
genero_pagos_columns = ["Genero","Ingresos"]

plt.figure(figsize=(6,4))

sns.barplot(x=genero_pagos.index,y=genero_pagos.values, palette=["pink","blue"])

plt.title("Cantidad de dinero que gana una mujer y un hombre")
plt.xlabel("Genero")
plt.ylabel("Ingresos")
plt.show()

```
